```python
import pandas as pd

reviews = pd.read_json("../data/preprocessed/reviews.jsonl", lines=True)
```

```python
long_sentences = []
for _, row in reviews.iterrows():
    for i, sent in enumerate(row.body_sentences):
        if len(sent) > 1000:
            long_sentences.append((_, i))
```

```python
lengths = reviews.body_sentences.apply(lambda lst: [len(sent) for sent in lst]).tolist()
lengths = [length for sentence_lengths in lengths for length in sentence_lengths]
print(f"Number of sentence lengths: {len(lengths)}")
print(f"Average length: {sum(lengths) / len(lengths)}")
```
