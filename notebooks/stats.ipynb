```python
import json
import itertools
from pathlib import Path
import numpy as np
```

```python
def fetch_results(dir: str):
    path = Path(f"../experiments/with_ndcg/{dir}/results.json")
    data = json.loads(path.read_text())
    return data


print(fetch_results("qwen8"))
```

```python
dirs = (
    "qwen8",
    "qwen8_add_title",
    "qwen8_add_abstract",
    "qwen8_add_title_prev_3",
    "qwen8_add_prev_3",
    "qwen8_add_prev_5",
    "qwen8_add_prev_7",
    "qwen8_add_prev_02",
)
metrics = ("average_hitrate_at_k", "average_recall_at_k", "average_ndcg_docs_at_k")
indexes = (9, 24, 49)
combinations = list(itertools.product(dirs, metrics, indexes))
```

```python
for dir, metric, idx in combinations:
    data = fetch_results(dir)
    print(f"{dir} - {metric} - {idx+1}: {data[metric][idx]:.4f}")
```

```python
rerank_results_dir = "../experiments/rerankers/results/qwen8_add_prev_02_bm25_gridsearch"
print(f"weights (BM25 / Cos) | hitrate@10 | hitrate@25 | hitrate@50 | recall@10 | recall@25 | recall@50 | ndcg@10 | ndcg@25 | ndcg@50")
for filename in sorted(Path(rerank_results_dir).glob("*.json")):
    data = json.loads(filename.read_text())
    weights = data["config"]["weights"]
    weight_label = f"{weights['score_bm25_scratch']:.1f} / {weights['metric']:.1f}"
    print(f"{weight_label} | {data['average_hitrate_at_k'][9]:.4f} | {data['average_hitrate_at_k'][24]:.4f} | {data['average_hitrate_at_k'][49]:.4f} | {data['average_recall_at_k'][9]:.4f} | {data['average_recall_at_k'][24]:.4f} | {data['average_recall_at_k'][49]:.4f} | {data['average_ndcg_docs_at_k'][9]:.4f} | {data['average_ndcg_docs_at_k'][24]:.4f} | {data['average_ndcg_docs_at_k'][49]:.4f}")
```
